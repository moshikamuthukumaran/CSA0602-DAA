```python
A=[[1,2],[3,4]]
B=[[5,6],[7,8]]
C=[[0,0],[0,0]]

for i in range(2):
    for j in range(2):
        for k in range(2):
            C[i][j]+=A[i][k]*B[k][j]

print(C)
```
