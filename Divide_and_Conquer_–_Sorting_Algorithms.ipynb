Merge Sort with Comparison Count

```python
arr = [12, 4, 78, 23, 45, 67, 89, 1]
comparisons = 0

def merge_sort(arr):
    global comparisons

    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    result = []
    i = j = 0

    while i < len(left) and j < len(right):
        comparisons += 1

        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


arr = merge_sort(arr)

print("Sorted Array:", ",".join(map(str, arr)))
print("Comparisons:", comparisons)
```

Quick Sort with Comparison Count

```python
arr = [38, 27, 43, 3, 9, 82, 10]
comparisons = 0

def partition(arr, low, high):
    global comparisons

    pivot = arr[high]
    i = low - 1

    for j in range(low, high):
        comparisons += 1

        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]

    return i + 1


def quick_sort(arr, low, high):
    if low < high:
        p = partition(arr, low, high)

        quick_sort(arr, low, p - 1)
        quick_sort(arr, p + 1, high)


quick_sort(arr, 0, len(arr) - 1)

print("Sorted Array:", ",".join(map(str, arr)))
print("Comparisons:", comparisons)
```

Hybrid Merge Sort with Insertion Sort

```python
arr = [12, 11, 13, 5, 6, 7, 3, 2]

THRESHOLD = 4

def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key

    return arr

def merge(left, right):
    result = []
    i = j = 0

    while i < len(left) and j < len(right):

        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result

def hybrid_merge_sort(arr):

    if len(arr) <= THRESHOLD:
        return insertion_sort(arr)

    mid = len(arr) // 2

    left = hybrid_merge_sort(arr[:mid])
    right = hybrid_merge_sort(arr[mid:])

    return merge(left, right)

result = hybrid_merge_sort(arr)

print("Sorted Array:", ",".join(map(str, result)))
```

Counting Inversions Using Merge Sort

```python
arr = [2, 4, 1, 3, 5]

def merge_sort(arr):
    if len(arr) <= 1:
        return arr, 0

    mid = len(arr) // 2

    left, inv_left = merge_sort(arr[:mid])
    right, inv_right = merge_sort(arr[mid:])

    merged = []
    i = j = 0
    inversions = inv_left + inv_right

    while i < len(left) and j < len(right):

        if left[i] <= right[j]:
            merged.append(left[i])
            i += 1

        else:
            merged.append(right[j])
            inversions += len(left) - i
            j += 1

    merged.extend(left[i:])
    merged.extend(right[j:])

    return merged, inversions

sorted_arr, inversions = merge_sort(arr)

print("Sorted:", ",".join(map(str, sorted_arr)))
print("Inversions:", inversions)
```

Quick Sort Recursion Depth Analysis

```python
arr = [10, 7, 8, 9, 1, 5]

max_depth = 0

def quick_sort(arr, low, high, depth):

    global max_depth

    if low < high:

        if depth > max_depth:
            max_depth = depth

        pivot = arr[high]
        i = low - 1

        for j in range(low, high):

            if arr[j] <= pivot:
                i += 1
                arr[i], arr[j] = arr[j], arr[i]

        arr[i + 1], arr[high] = arr[high], arr[i + 1]

        p = i + 1

        quick_sort(arr, low, p - 1, depth + 1)
        quick_sort(arr, p + 1, high, depth + 1)

quick_sort(arr, 0, len(arr) - 1, 1)

print("Sorted Array:", ",".join(map(str, arr)))
print("Max Depth:", max_depth)
```

Space Complexity Comparison — Merge Sort vs Quick Sort

```python
import sys

arr = [5, 3, 8, 4, 2]

def merge_sort(arr):

    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    result = []
    i = j = 0

    while i < len(left) and j < len(right):

        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result

def quick_sort(arr, low, high):

    if low < high:

        pivot = arr[high]
        i = low - 1

        for j in range(low, high):

            if arr[j] <= pivot:
                i += 1
                arr[i], arr[j] = arr[j], arr[i]

        arr[i + 1], arr[high] = arr[high], arr[i + 1]

        p = i + 1

        quick_sort(arr, low, p - 1)
        quick_sort(arr, p + 1, high)


merge_result = merge_sort(arr.copy())

quick_result = arr.copy()
quick_sort(quick_result, 0, len(quick_result) - 1)

print("Sorted:", ",".join(map(str, merge_result)))

print("Merge Sort auxiliary space: O(n)")
print("Quick Sort auxiliary space: O(log n) average")
```

K-th Smallest Element Using Quick Sort Partition

```python
arr = [7, 10, 4, 3, 20, 15]
k = 3

def partition(arr, low, high):

    pivot = arr[high]
    i = low - 1

    for j in range(low, high):

        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]

    return i + 1

def kth_smallest(arr, low, high, k):

    if low <= high:

        p = partition(arr, low, high)

        if p == k:
            return arr[p]

        elif p > k:
            return kth_smallest(arr, low, p - 1, k)

        else:
            return kth_smallest(arr, p + 1, high, k)

result = kth_smallest(arr, 0, len(arr) - 1, k - 1)

print("K-th Smallest Element:", result)
```
