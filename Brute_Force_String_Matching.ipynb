Basic Pattern Search

```python
def brute_force(text, pattern):
    positions = []
    comparisons = 0

    for i in range(len(text) - len(pattern) + 1):
        j = 0

        while j < len(pattern):
            comparisons += 1
            if text[i + j] != pattern[j]:
                break
            j += 1

        if j == len(pattern):
            positions.append(i)

    print("Positions:", positions)
    print("Total comparisons:", comparisons)


brute_force("AABAACAADAABAABA", "AABA")
```

Find All Occurrences

```python
def find_all(text, pattern):
    positions = []

    for i in range(len(text) - len(pattern) + 1):
        j = 0

        while j < len(pattern) and text[i + j] == pattern[j]:
            j += 1

        if j == len(pattern):
            positions.append(i)

    print("Occurrences at positions:", positions)


find_all("BANANABANANA", "ANA")
```

Comparison Table

```python
def comparison_table(text, pattern):
    m = len(pattern)

    for i in range(len(text) - m + 1):
        comparisons = 0
        matched = True

        for j in range(m):
            comparisons += 1

            if text[i + j] != pattern[j]:
                matched = False
                break

        result = "Match" if matched else "Mismatch"
        print("Shift:", i, "| Comparisons:", comparisons, "|", result)


comparison_table("MISSISSIPPI", "ISSI")
```

Comparison Count

```python
def count_comparisons(text, pattern):
    comparisons = 0
    matches = 0
    mismatches = 0

    for i in range(len(text) - len(pattern) + 1):
        for j in range(len(pattern)):
            comparisons += 1

            if text[i + j] == pattern[j]:
                matches += 1
            else:
                mismatches += 1
                break

    print("Total comparisons:", comparisons)
    print("Total matches:", matches)
    print("Total mismatches:", mismatches)


count_comparisons("ABABABABAB", "ABAB")
```

Best, Average and Worst Case

```python
def brute_force_analysis(text, pattern):
    comparisons = 0

    for i in range(len(text) - len(pattern) + 1):
        for j in range(len(pattern)):
            comparisons += 1

            if text[i + j] != pattern[j]:
                break

            if j == len(pattern) - 1:
                print("Pattern found at position:", i)

    print("Comparisons:", comparisons)
    print("Best Case: O(n)")
    print("Average Case: O(nm)")
    print("Worst Case: O(nm)")


brute_force_analysis("AAAAAAAAAB", "AAAAB")
```

First Occurrence

```python
def first_occurrence(text, pattern):
    comparisons = 0

    for i in range(len(text) - len(pattern) + 1):
        j = 0

        while j < len(pattern):
            comparisons += 1

            if text[i + j] != pattern[j]:
                break

            j += 1

        if j == len(pattern):
            print("First occurrence:", i)
            print("Comparisons:", comparisons)
            return

    print("Pattern not found")
    print("Comparisons:", comparisons)


first_occurrence("COMPUTERSCIENCE", "SCI")
```

Case-Insensitive Search

```python
def case_insensitive_search(text, pattern):
    text = text.lower()
    pattern = pattern.lower()

    for i in range(len(text) - len(pattern) + 1):
        if text[i:i + len(pattern)] == pattern:
            print("Pattern found at position:", i)
            return

    print("Pattern not found")


case_insensitive_search("DataStructuresAndAlgorithms", "ALGORITHMS")
```

Successful vs Unsuccessful Search

```python
def search_count(text, pattern):
    comparisons = 0

    for i in range(len(text) - len(pattern) + 1):
        for j in range(len(pattern)):
            comparisons += 1

            if text[i + j] != pattern[j]:
                break
        else:
            return comparisons

    return comparisons


text = "PROGRAMMINGLAB"

print("LAB comparisons:", search_count(text, "LAB"))
print("TEST comparisons:", search_count(text, "TEST"))
```

Alignment Display

```python
def show_alignments(text, pattern):
    occurrence = []

    for i in range(len(text) - len(pattern) + 1):
        matched = True

        for j in range(len(pattern)):
            if text[i + j] != pattern[j]:
                matched = False
                break

        print("Alignment", i + 1, ":", "Match" if matched else "Mismatch")

        if matched:
            occurrence.append(i)

    print("Pattern positions:", occurrence)


show_alignments("ABCDABCABCDA", "ABCDA")
```

Complete Brute Force Analysis

```python
def brute_force_complete(text, pattern):
    positions = []
    comparisons = 0

    for i in range(len(text) - len(pattern) + 1):
        print("Shift", i, end=": ")

        matched = True

        for j in range(len(pattern)):
            comparisons += 1

            if text[i + j] != pattern[j]:
                matched = False
                print("Mismatch")
                break

        if matched:
            positions.append(i)
            print("Match")

    print("\nOccurrences:", positions)
    print("Total comparisons:", comparisons)

    print("\nComplexity:")
    print("Best Case: O(n)")
    print("Worst Case: O(nm)")
    print("Space Complexity: O(1)")


brute_force_complete(
    "TTATAGATCTCGTATTCTTTATAGATCTCCTATTCTT",
    "TATCTT"
)
```
