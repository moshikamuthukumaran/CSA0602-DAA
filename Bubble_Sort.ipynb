Bubble Sort – Exam Result Slip

```python
def optimized_bubble_sort(arr):
    a = arr.copy()
    passes = 0

    for i in range(len(a)):
        swapped = False
        passes += 1

        for j in range(len(a) - 1 - i):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swapped = True

        if not swapped:
            break

    return a, passes


rolls = [101, 102, 104, 103, 105, 107, 106, 108]

result, passes = optimized_bubble_sort(rolls)

print("Sorted:", result)
print("Passes:", passes)
```

Bubble Sort – Traffic Signal Queue

```python
def bubble_sort_queue(queue):
    priority = {
        "ambulance": 1,
        "bus": 2,
        "car": 3
    }

    a = queue.copy()

    for i in range(len(a)):
        for j in range(len(a) - 1 - i):
            if priority[a[j]] > priority[a[j + 1]]:
                a[j], a[j + 1] = a[j + 1], a[j]

    return a


queue = ["car", "car", "bus"]
queue.append("ambulance")

print("Priority queue:", bubble_sort_queue(queue))
```

Bubble Sort – Visualization

```python
def bubble_sort_with_frames(arr):
    a = arr.copy()
    frames = [a.copy()]

    for i in range(len(a)):
        swapped = False

        for j in range(len(a) - 1 - i):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swapped = True

        frames.append(a.copy())

        if not swapped:
            break

    return frames


frames = bubble_sort_with_frames([5, 1, 4, 2, 8])

for i, frame in enumerate(frames):
    print("Pass", i, ":", frame)
```

Bubble Sort – Sensor Alerts

```python
def bubble_sort_plain(arr):
    a = arr.copy()
    comparisons = 0

    for i in range(len(a)):
        for j in range(len(a) - 1 - i):
            comparisons += 1

            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]

    return a, comparisons


def bubble_sort_optimized(arr):
    a = arr.copy()
    comparisons = 0

    for i in range(len(a)):
        swapped = False

        for j in range(len(a) - 1 - i):
            comparisons += 1

            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swapped = True

        if not swapped:
            break

    return a, comparisons


alerts = [2, 1, 3, 2, 1, 4, 3, 2, 5, 1, 2, 3, 4, 1, 2]

r1, c1 = bubble_sort_plain(alerts)
r2, c2 = bubble_sort_optimized(alerts)

print("Sorted:", r2)
print("Plain comparisons:", c1)
print("Optimized comparisons:", c2)
```
