Dijkstra's Algorithm

Q1. Shortest path from vertex 0 to all vertices


```python
import heapq

def dijkstra(n, edges, src):
    g = [[] for _ in range(n)]
    for u, v, w in edges:
        g[u].append((v, w))

    dist = [float('inf')] * n
    parent = [-1] * n
    dist[src] = 0
    pq = [(0, src)]

    while pq:
        d, u = heapq.heappop(pq)
        if d > dist[u]:
            continue

        for v, w in g[u]:
            if d + w < dist[v]:
                dist[v] = d + w
                parent[v] = u
                heapq.heappush(pq, (dist[v], v))

    return dist, parent

def path(parent, v):
    p = []
    while v != -1:
        p.append(v)
        v = parent[v]
    return p[::-1]

edges = [(0,1,4),(0,2,1),(2,1,2),(1,3,1),(2,3,5),(3,4,3)]

dist, parent = dijkstra(5, edges, 0)

print("Distances from 0:", dist)
for v in range(1, 5):
    print("Path to", v, ":", " -> ".join(map(str, path(parent, v))))
```

Q2. Shortest path from source 0 to destination 4

```python
import heapq

def shortest_path(n, edges, src, dst):
    g = [[] for _ in range(n)]

    for u, v, w in edges:
        g[u].append((v, w))

    dist = [float('inf')] * n
    parent = [-1] * n
    dist[src] = 0
    pq = [(0, src)]

    while pq:
        d, u = heapq.heappop(pq)

        if d > dist[u]:
            continue

        for v, w in g[u]:
            if d + w < dist[v]:
                dist[v] = d + w
                parent[v] = u
                heapq.heappush(pq, (dist[v], v))

    p = []
    v = dst
    while v != -1:
        p.append(v)
        v = parent[v]

    return dist[dst], p[::-1]

edges = [(0,1,10),(0,2,3),(1,3,2),(2,1,4),
         (2,3,8),(2,4,2),(3,4,5)]

d, p = shortest_path(5, edges, 0, 4)

print("Shortest distance:", d)
print("Path:", " -> ".join(map(str, p)))
```

Kruskal's Algorithm

Q1. Basic MST

```python
def kruskal(n, edges):
    parent = list(range(n))

    def find(x):
        while parent[x] != x:
            x = parent[x]
        return x

    mst = []
    total = 0

    for u, v, w in sorted(edges, key=lambda x: x[2]):
        a, b = find(u), find(v)

        if a != b:
            parent[a] = b
            mst.append((u, v, w))
            total += w

    return mst, total

edges = [(0,1,10),(0,2,6),(0,3,5),(1,3,15),(2,3,4)]

mst, total = kruskal(4, edges)

print("Edges in MST:", mst)
print("Total weight of MST:", total)
```

Q2. Kruskal with tie-breaking

```python
def kruskal(n, edges):
    parent = list(range(n))

    def find(x):
        if parent[x] != x:
            parent[x] = find(parent[x])
        return parent[x]

    mst = []
    total = 0

    edges.sort(key=lambda x: (x[2], x[0], x[1]))

    for u, v, w in edges:
        a, b = find(u), find(v)

        if a != b:
            parent[a] = b
            mst.append((u, v, w))
            total += w

    return mst, total

edges = [(0,1,2),(0,3,6),(1,2,3),(1,3,8),
         (1,4,5),(2,4,7),(3,4,9)]

mst, total = kruskal(5, edges)

print("Edges in MST:", mst)
print("Total weight of MST:", total)
```

Q4. Disconnected Graph – Spanning Forest

```python
def kruskal(n, edges):
    parent = list(range(n))

    def find(x):
        if parent[x] != x:
            parent[x] = find(parent[x])
        return parent[x]

    for u, v, w in sorted(edges, key=lambda x: x[2]):
        a, b = find(u), find(v)

        if a != b:
            parent[a] = b

    return parent

def spanning_forest(n, edges):
    parent = list(range(n))

    def find(x):
        if parent[x] != x:
            parent[x] = find(parent[x])
        return parent[x]

    groups = {}
    total = 0

    for u, v, w in sorted(edges, key=lambda x: x[2]):
        a, b = find(u), find(v)

        if a != b:
            parent[a] = b

    for i in range(n):
        root = find(i)
        groups.setdefault(root, [])

    for u, v, w in sorted(edges, key=lambda x: x[2]):
        a, b = find(u), find(v)

        if a != b:
            parent[a] = b

    # Rebuild MST
    parent = list(range(n))
    result = []

    for u, v, w in sorted(edges, key=lambda x: x[2]):
        a, b = find(u), find(v)
        if a != b:
            parent[a] = b
            result.append((u, v, w))
            total += w

    return result, total

edges = [(0,1,3),(0,2,5),(1,2,2),
         (3,4,4),(3,5,6),(4,5,1),(6,6,0)]

mst, total = spanning_forest(7, edges)

print("Spanning forest:", mst)
print("Total MST weight:", total)
```

Q5. MST Sensitivity Analysis

```python
def kruskal(n, edges):
    parent = list(range(n))

    def find(x):
        if parent[x] != x:
            parent[x] = find(parent[x])
        return parent[x]

    mst = []

    for e in sorted(edges, key=lambda x: x[2]):
        u, v, w = e
        a, b = find(u), find(v)

        if a != b:
            parent[a] = b
            mst.append(e)

    return mst

edges = [(0,1,1),(0,2,5),(1,2,3),(1,3,4),
         (2,3,2),(2,4,6),(3,4,7),(0,4,9)]

mst = kruskal(5, edges)
weight = sum(e[2] for e in mst)

non_mst = [e for e in edges if e not in mst]

print("MST edges:", mst)
print("Weight:", weight)
print("Non-MST edges:", non_mst)

best = None

for edge in non_mst:
    new_edges = [e for e in edges if e != edge]
    new_mst = kruskal(5, new_edges + [edge])
    new_weight = sum(e[2] for e in new_mst)

    if best is None or new_weight < best[0]:
        best = (new_weight, edge, new_mst)

print("Best swap: add", best[1])
print("New MST weight:", best[0])
```

Prim's Algorithm

Q1. Prim's starting from vertex 0

```python
def prim(n, edges, start):
    g = [[] for _ in range(n)]

    for u, v, w in edges:
        g[u].append((v, w))
        g[v].append((u, w))

    used = [False] * n
    used[start] = True
    mst = []
    total = 0

    for _ in range(n - 1):
        best = None

        for u in range(n):
            if used[u]:
                for v, w in g[u]:
                    if not used[v] and (best is None or w < best[2]):
                        best = (u, v, w)

        u, v, w = best
        used[v] = True
        mst.append(best)
        total += w
        print("Add:", best)

    return mst, total

edges = [(0,1,10),(0,2,6),(0,3,5),(1,3,15),(2,3,4)]

mst, total = prim(4, edges, 0)

print("MST edges:", mst)
print("Total weight:", total)
```

Q2. Prim's with key array

```python
def prim(n, edges, start):
    graph = [[float('inf')] * n for _ in range(n)]

    for u, v, w in edges:
        graph[u][v] = graph[v][u] = w

    key = [float('inf')] * n
    parent = [-1] * n
    used = [False] * n

    key[start] = 0
    print("Initial key[]:", key)

    for _ in range(n):
        u = min((i for i in range(n) if not used[i]),
                key=lambda i: key[i])

        used[u] = True

        for v in range(n):
            if not used[v] and graph[u][v] < key[v]:
                key[v] = graph[u][v]
                parent[v] = u

        print("After picking", u, ":", key)

    mst = []
    for i in range(n):
        if parent[i] != -1:
            mst.append((parent[i], i, key[i]))

    return mst, sum(key[i] for i in range(n))

edges = [(0,1,2),(0,3,6),(1,2,3),(1,3,8),
         (1,4,5),(2,4,7),(3,4,9)]

mst, total = prim(5, edges, 2)

print("MST edges:", mst)
print("Weight:", total)
```

Borůvka's Algorithm

Q1. Cheapest edge for each component

```python
def boruvka(n, edges):
    parent = list(range(n))

    def find(x):
        if parent[x] != x:
            parent[x] = find(parent[x])
        return parent[x]

    mst = []
    total = 0
    components = n

    while components > 1:
        cheapest = [None] * n

        for u, v, w in edges:
            a, b = find(u), find(v)

            if a != b:
                if cheapest[a] is None or w < cheapest[a][2]:
                    cheapest[a] = (u, v, w)
                if cheapest[b] is None or w < cheapest[b][2]:
                    cheapest[b] = (u, v, w)

        for edge in cheapest:
            if edge:
                u, v, w = edge
                a, b = find(u), find(v)

                if a != b:
                    parent[a] = b
                    mst.append(edge)
                    total += w
                    components -= 1

    return mst, total

edges = [(0,1,10),(0,2,6),(0,3,5),(1,3,15),(2,3,4)]

mst, total = boruvka(4, edges)

print("MST edges:", mst)
print("Weight:", total)
```

Q2. Borůvka phases

```python
def boruvka(n, edges):
    parent = list(range(n))
    components = n
    total = 0

    def find(x):
        if parent[x] != x:
            parent[x] = find(parent[x])
        return parent[x]

    phase = 0

    while components > 1:
        phase += 1
        cheapest = [None] * n

        for u, v, w in edges:
            a, b = find(u), find(v)

            if a != b:
                if cheapest[a] is None or w < cheapest[a][2]:
                    cheapest[a] = (u, v, w)
                if cheapest[b] is None or w < cheapest[b][2]:
                    cheapest[b] = (u, v, w)

        added = []

        for e in cheapest:
            if e:
                u, v, w = e
                a, b = find(u), find(v)

                if a != b:
                    parent[a] = b
                    added.append(e)
                    total += w
                    components -= 1

        print("Phase", phase, "edges added:", added)

    return total

edges = [(0,1,2),(0,3,6),(1,2,3),(1,3,8),
         (1,4,5),(2,4,7),(3,4,9)]

print("Total weight:", boruvka(5, edges))
```
