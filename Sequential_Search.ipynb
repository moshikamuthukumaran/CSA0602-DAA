Basic Sequential Search

```python
def sequential_search(arr, key):
    comparisons = 0

    for i in range(len(arr)):
        comparisons += 1

        if arr[i] == key:
            print("Element found at position", i + 1)
            print("Number of comparisons =", comparisons)
            return

    print("Element not found")
    print("Number of comparisons =", comparisons)


arr = [12, 25, 8, 45, 32, 19, 50]
sequential_search(arr, 32)
```

Unsuccessful Search

```python
def sequential_search(arr, key):
    comparisons = 0

    for x in arr:
        comparisons += 1

        if x == key:
            print("Element found")
            print("Number of comparisons =", comparisons)
            return

    print("Element not found")
    print("Number of comparisons =", comparisons)


arr = [5, 10, 15, 20, 25, 30, 35]
sequential_search(arr, 18)
```

First Occurrence Search

```python
def first_occurrence(arr, key):
    for i in range(len(arr)):
        if arr[i] == key:
            print("First occurrence at position", i + 1)
            return

    print("Element not found")


arr = [10, 25, 15, 25, 30, 25, 40]
first_occurrence(arr, 25)
```

Find All Occurrences

```python
def all_occurrences(arr, key):
    positions = []

    for i in range(len(arr)):
        if arr[i] == key:
            positions.append(i + 1)

    print("Occurrences at positions:")
    print(*positions, sep=", ")
    print("Total occurrences =", len(positions))


arr = [7, 12, 7, 25, 18, 7, 30, 7]
all_occurrences(arr, 7)
```

Comparison and Match Count

```python
def search_count(arr, key):
    comparisons = 0
    matches = 0
    mismatches = 0

    for x in arr:
        comparisons += 1

        if x == key:
            matches += 1
            break
        else:
            mismatches += 1

    print("Total comparisons =", comparisons)
    print("Total matches =", matches)
    print("Total mismatches =", mismatches)


arr = [3, 6, 9, 12, 15, 18, 21]
search_count(arr, 15)
```

Sentinel Search

```python
def sentinel_search(arr, key):
    a = arr.copy()
    n = len(a)

    a.append(key)
    i = 0
    comparisons = 0

    while a[i] != key:
        comparisons += 1
        i += 1

    if i < n:
        comparisons += 1
        print("Position found:", i + 1)
    else:
        print("Element not found")

    print("Comparison count:", comparisons)


arr = [14, 9, 22, 35, 18, 41, 27]
sentinel_search(arr, 18)
```

Register Number Search

```python
def search_register(registers, key):
    for i in range(len(registers)):
        if registers[i] == key:
            print("Register Number found at position", i + 1)
            return

    print("Register Number not found")


registers = [101, 102, 103, 104, 105, 106]
search_register(registers, 104)
```

String Search

```python
def search_name(names, key):
    for i in range(len(names)):
        if names[i] == key:
            print("Name found at position", i + 1)
            return

    print("Name not found")


names = ["Anu", "Bala", "Charan", "Deepa", "Esha", "Farhan"]
search_name(names, "Deepa")
```

2-D Matrix Search

```python
def matrix_search(matrix, key):
    for i in range(len(matrix)):
        for j in range(len(matrix[i])):
            if matrix[i][j] == key:
                print("Element found at Row", i + 1, "Column", j + 1)
                return

    print("Element not found")


matrix = [
    [12, 8, 15],
    [5, 18, 27],
    [9, 11, 24]
]

matrix_search(matrix, 24)
```

Search Performance Analysis

```python
def sequential_search(arr, key):
    comparisons = 0

    for i in range(len(arr)):
        comparisons += 1
        print("Comparison", comparisons, ":", arr[i], "with", key)

        if arr[i] == key:
            print("Found at position", i + 1)
            print("Comparisons =", comparisons)
            return comparisons

    print("Not found")
    print("Comparisons =", comparisons)
    return comparisons


arr = [45, 23, 67, 12, 89, 34, 56, 78, 90, 11,
       29, 73, 18, 64, 37]

keys = [73, 18, 100]

for key in keys:
    print("\nSearching for:", key)
    sequential_search(arr, key)

print("\nPerformance Analysis")
print("Best Case: O(1)")
print("Average Case: O(n)")
print("Worst Case: O(n)")
print("Space Complexity: O(1)")
```
