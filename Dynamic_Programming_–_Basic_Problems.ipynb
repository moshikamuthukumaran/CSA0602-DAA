Dice Throw

```python
def dice_ways(n, m, target):
    dp = [0] * (target + 1)
    dp[0] = 1

    for _ in range(n):
        new_dp = [0] * (target + 1)

        for s in range(target + 1):
            for face in range(1, m + 1):
                if s >= face:
                    new_dp[s] += dp[s - face]

        dp = new_dp

    return dp[target]


tests = [
    (3, 6, 8),
    (2, 6, 7),
    (4, 4, 10),
    (3, 6, 12),
    (5, 6, 15),
    (3, 5, 9),
    (2, 10, 11),
    (2, 6, 7),
    (3, 6, 8),
    (4, 6, 10),
    (2, 4, 5)
]

for n, m, target in tests:
    print("Dice:", n, "Faces:", m, "Target:", target)
    print("Number of ways =", dice_ways(n, m, target))
    print()
```

Binomial Coefficient

```python
def binomial(n, r):
    if r < 0 or r > n:
        return 0

    dp = [[0] * (r + 1) for _ in range(n + 1)]

    for i in range(n + 1):
        dp[i][0] = 1

    for i in range(1, n + 1):
        for j in range(1, min(i, r) + 1):
            dp[i][j] = dp[i - 1][j - 1] + dp[i - 1][j]

    return dp[n][r]


tests = [
    (3, 7),   # Dice question: not used here
    (5, 2),
    (8, 3),
    (10, 4),
    (12, 5),
    (15, 6),
    (10, 3),
    (12, 4),
    (15, 5),
    (20, 6),
    (8, 3)
]

for i, (n, r) in enumerate(tests):
    if i == 0:
        continue

    print("n =", n, "r =", r)
    print("C(", n, ",", r, ") =", binomial(n, r))
    print()
```

Assembly Line Scheduling

```python
def assembly_line(entry, exit, line1, line2, transfer1, transfer2):
    n = len(line1)

    dp1 = [0] * n
    dp2 = [0] * n

    dp1[0] = entry[0] + line1[0]
    dp2[0] = entry[1] + line2[0]

    for i in range(1, n):
        dp1[i] = min(
            dp1[i - 1] + line1[i],
            dp2[i - 1] + transfer2[i - 1] + line1[i]
        )

        dp2[i] = min(
            dp2[i - 1] + line2[i],
            dp1[i - 1] + transfer1[i - 1] + line2[i]
        )

    return min(
        dp1[n - 1] + exit[0],
        dp2[n - 1] + exit[1]
    )


tests = [
    (
        [10, 12], [18, 7],
        [4, 5, 3, 2], [2, 10, 1, 4],
        [7, 4, 5], [9, 2, 8]
    ),

    (
        [8, 10], [12, 15],
        [5, 6, 4], [7, 3, 5],
        [2, 1], [3, 2]
    ),

    (
        [5, 6], [4, 5],
        [3, 4, 5], [4, 3, 6],
        [2, 1], [1, 2]
    ),

    (
        [7, 9], [8, 6],
        [4, 3, 5, 2], [5, 2, 4, 3],
        [1, 2, 1], [2, 1, 2]
    ),

    (
        [6, 7], [5, 4],
        [2, 4, 3], [3, 2, 5],
        [1, 2], [2, 1]
    )
]


for i, test in enumerate(tests, 1):
    result = assembly_line(*test)

    print("Question", i)
    print("Minimum Production Time =", result)
    print()
```
