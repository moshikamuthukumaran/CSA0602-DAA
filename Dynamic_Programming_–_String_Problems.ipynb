Longest Palindromic Subsequence

DNA Palindrome Motif Detection

```python
def lps_length(s):
    n = len(s)

    if n == 0:
        return 0

    dp = [[0] * n for _ in range(n)]

    for i in range(n):
        dp[i][i] = 1

    for length in range(2, n + 1):
        for i in range(n - length + 1):
            j = i + length - 1

            if s[i] == s[j]:
                dp[i][j] = 2 + (dp[i + 1][j - 1] if length > 2 else 0)
            else:
                dp[i][j] = max(dp[i + 1][j], dp[i][j - 1])

    return dp[0][n - 1]


print("LPS Length =", lps_length("bbbab"))
```

Reconstructing the Actual Subsequence

```python
def reconstruct_lps(s):
    n = len(s)

    if n == 0:
        return ""

    dp = [[0] * n for _ in range(n)]

    for i in range(n):
        dp[i][i] = 1

    for length in range(2, n + 1):
        for i in range(n - length + 1):
            j = i + length - 1

            if s[i] == s[j]:
                dp[i][j] = 2 + (dp[i + 1][j - 1]
                                 if length > 2 else 0)
            else:
                dp[i][j] = max(dp[i + 1][j],
                                dp[i][j - 1])

    i, j = 0, n - 1
    left = ""
    right = ""

    while i <= j:
        if i == j:
            left += s[i]
            break

        if s[i] == s[j]:
            left += s[i]
            right = s[j] + right
            i += 1
            j -= 1

        elif dp[i + 1][j] >= dp[i][j - 1]:
            i += 1
        else:
            j -= 1

    return left + right


print("Longest Palindromic Subsequence =",
      reconstruct_lps("bbbab"))
```

Memory-Optimized LPS

```python
def lps_length_optimized(s):
    n = len(s)

    if n == 0:
        return 0

    dp = [1] * n

    for i in range(n - 2, -1, -1):
        prev = 0

        for j in range(i + 1, n):
            temp = dp[j]

            if s[i] == s[j]:
                dp[j] = prev + 2
            else:
                dp[j] = max(dp[j], dp[j - 1])

            prev = temp

    return dp[n - 1]


print("Optimized LPS Length =",
      lps_length_optimized("bbbab"))
```

Password Weakness Checker

```python
def is_weak_password(password, threshold):
    if len(password) == 0:
        return False

    ratio = lps_length_optimized(password) / len(password)

    return ratio > threshold


print("Weak Password =",
      is_weak_password("bbbab", 0.6))
```

LPS vs LCS Reformulation

```python
def lps_length(s):
    n = len(s)

    if n == 0:
        return 0

    dp = [[0] * n for _ in range(n)]

    for i in range(n):
        dp[i][i] = 1

    for length in range(2, n + 1):
        for i in range(n - length + 1):
            j = i + length - 1

            if s[i] == s[j]:
                dp[i][j] = 2 + (dp[i + 1][j - 1]
                                 if length > 2 else 0)
            else:
                dp[i][j] = max(dp[i + 1][j],
                                dp[i][j - 1])

    return dp[0][n - 1]


def lcs_length(a, b):
    n = len(a)
    m = len(b)

    dp = [[0] * (m + 1) for _ in range(n + 1)]

    for i in range(1, n + 1):
        for j in range(1, m + 1):

            if a[i - 1] == b[j - 1]:
                dp[i][j] = dp[i - 1][j - 1] + 1
            else:
                dp[i][j] = max(dp[i - 1][j],
                                dp[i][j - 1])

    return dp[n][m]


s = "bbbab"

print("LPS Length =", lps_length(s))
print("LCS Length =", lcs_length(s, s[::-1]))
print("Both are equal:", lps_length(s) == lcs_length(s, s[::-1]))
```

All Possible Segmentations

```python
def all_segmentations(s, dictionary):
    words = set(dictionary)
    memo = {}

    def solve(start):
        if start == len(s):
            return [""]

        if start in memo:
            return memo[start]

        result = []

        for end in range(start + 1, len(s) + 1):
            if s[start:end] in words:
                for rest in solve(end):
                    result.append(s[start:end] +
                                  (" " + rest if rest else ""))

        memo[start] = result
        return result

    return solve(0)


result = all_segmentations(
    "catsanddog",
    ["cats", "cat", "and", "sand", "dog"]
)

print(result)
```

Word Wrap Problem

Justifying Text for a Print Newsletter Column

```python
def min_wrap_cost(words, width):
    n = len(words)
    dp = [float('inf')] * (n + 1)
    dp[n] = 0

    for i in range(n - 1, -1, -1):
        length = 0

        for j in range(i, n):
            length += words[j]

            if j > i:
                length += 1

            if length > width:
                break

            cost = 0 if j == n - 1 else (width - length) ** 2
            dp[i] = min(dp[i], cost + dp[j + 1])

    return dp[0]


print("Minimum Wrap Cost =", min_wrap_cost([3, 2, 2, 5], 6))
```
