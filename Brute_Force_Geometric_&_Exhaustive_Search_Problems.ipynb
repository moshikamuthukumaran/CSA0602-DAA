Closest Pair of Points — Brute Force

```python
import math
points = [(1, 2), (4, 5), (7, 8), (3, 1)]

min_distance = float('inf')
closest_pair = None

for i in range(len(points)):
    for j in range(i + 1, len(points)):
        x1, y1 = points[i]
        x2, y2 = points[j]

        distance = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)

        if distance < min_distance:
            min_distance = distance
            closest_pair = (points[i], points[j])

print("Closest pair:", closest_pair[0], "-", closest_pair[1])
print("Minimum distance:", min_distance)
```

Closest Pair + Convex Hull Brute Force

```python
import math

def distance(p1, p2):
    return math.sqrt((p1[0] - p2[0]) ** 2 + (p1[1] - p2[1]) ** 2)

def closest_pair(points):
    min_dist = float('inf')
    pair = None

    for i in range(len(points)):
        for j in range(i + 1, len(points)):
            d = distance(points[i], points[j])

            if d < min_dist:
                min_dist = d
                pair = (points[i], points[j])

    return pair, min_dist

points = [(1, 2), (4, 5), (7, 8), (3, 1)]

pair, min_dist = closest_pair(points)

print("Closest pair:", pair[0], "-", pair[1])
print("Minimum distance:", min_dist)

print("Time Complexity: O(n^2)")

points = [
    (10, 0),
    (11, 5),
    (5, 3),
    (9, 3.5),
    (15, 3),
    (12.5, 7),
    (6, 6.5),
    (7.5, 4.5)
]

def orientation(a, b, c):
    value = (b[0] - a[0]) * (c[1] - a[1]) - \
            (b[1] - a[1]) * (c[0] - a[0])

    if value > 0:
        return 1
    elif value < 0:
        return -1
    return 0

hull = []

for i in range(len(points)):
    for j in range(i + 1, len(points)):
        pos = neg = False

        for k in range(len(points)):
            if k == i or k == j:
                continue

            o = orientation(points[i], points[j], points[k])

            if o > 0:
                pos = True
            elif o < 0:
                neg = True

        if not (pos and neg):
            hull.append(points[i])
            hull.append(points[j])

hull = list(set(hull))

print("Convex Hull Points:")

for point in hull:
    print(point)
```

```python
def orientation(a, b, c):
    value = (b[0] - a[0]) * (c[1] - a[1]) - \
            (b[1] - a[1]) * (c[0] - a[0])

    if value > 0:
        return 1
    elif value < 0:
        return -1
    return 0

def convex_hull(points):
    hull = []

    for i in range(len(points)):
        for j in range(i + 1, len(points)):
            left = right = False

            for k in range(len(points)):
                if k == i or k == j:
                    continue

                o = orientation(points[i], points[j], points[k])

                if o > 0:
                    left = True
                elif o < 0:
                    right = True

            if not (left and right):
                hull.append(points[i])
                hull.append(points[j])

    return list(set(hull))


points = [(0, 0), (1, 1), (2, 2), (0, 2), (2, 0)]

hull = convex_hull(points)

print("Convex Hull Points:")
for point in hull:
    print(point)
```

Travelling Salesman Problem — Exhaustive Search

```python
import math
from itertools import permutations

def distance(p1, p2):
    return math.sqrt(
        (p1[0] - p2[0]) ** 2 +
        (p1[1] - p2[1]) ** 2
    )

def tsp(points):
    start = points[0]
    remaining = points[1:]

    min_distance = float('inf')
    best_path = None

    for perm in permutations(remaining):
        path = [start] + list(perm) + [start]

        total = 0

        for i in range(len(path) - 1):
            total += distance(path[i], path[i + 1])

        if total < min_distance:
            min_distance = total
            best_path = path

    return min_distance, best_path


test_cases = [
    [(1, 2), (4, 5), (7, 1), (3, 6)],
    [(2, 4), (8, 1), (1, 7), (6, 3), (5, 9)]
]

for i, points in enumerate(test_cases, 1):
    distance_value, path = tsp(points)

    print("Test Case", i)
    print("Shortest Distance:", distance_value)
    print("Shortest Path:", path)
    print()
```

Assignment Problem — Exhaustive Search

```python
from itertools import permutations

def total_cost(assignment, cost_matrix):
    total = 0

    for worker, task in enumerate(assignment):
        total += cost_matrix[worker][task]

    return total

def assignment_problem(cost_matrix):
    n = len(cost_matrix)

    min_cost = float('inf')
    best_assignment = None

    for assignment in permutations(range(n)):
        cost = total_cost(assignment, cost_matrix)

        if cost < min_cost:
            min_cost = cost
            best_assignment = assignment

    return best_assignment, min_cost

test_cases = [
    [
        [3, 10, 7],
        [8, 5, 12],
        [4, 6, 9]
    ],
    [
        [15, 9, 4],
        [8, 7, 18],
        [6, 12, 11]
    ]
]

for i, matrix in enumerate(test_cases, 1):

    assignment, cost = assignment_problem(matrix)

    print("Test Case", i)
    print("Optimal Assignment:")

    for worker, task in enumerate(assignment):
        print(f"(worker {worker + 1}, task {task + 1})")

    print("Total Cost:", cost)
    print()
```

0-1 Knapsack — Exhaustive Search

```python
from itertools import combinations

def total_value(items, values):
    total = 0

    for item in items:
        total += values[item]

    return total

def is_feasible(items, weights, capacity):
    total_weight = 0

    for item in items:
        total_weight += weights[item]

    return total_weight <= capacity

def knapsack(weights, values, capacity):
    n = len(weights)

    best_items = []
    best_value = 0

    for r in range(n + 1):
        for items in combinations(range(n), r):

            if is_feasible(items, weights, capacity):

                value = total_value(items, values)

                if value > best_value:
                    best_value = value
                    best_items = list(items)

    return best_items, best_value


test_cases = [
    ([2, 3, 1], [4, 5, 3], 4),
    ([1, 2, 3, 4], [2, 4, 6, 3], 6)
]

for i, (weights, values, capacity) in enumerate(test_cases, 1):

    items, value = knapsack(weights, values, capacity)

    print("Test Case", i)
    print("Optimal Selection:", items)
    print("Total Value:", value)
    print()
```
