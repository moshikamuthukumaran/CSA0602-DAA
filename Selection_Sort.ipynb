Selection Sort – Trophy Ranking

```python
def top_k_scores(scores, k):
    a = scores.copy()

    for i in range(min(k, len(a))):
        max_index = i

        for j in range(i + 1, len(a)):
            if a[j] > a[max_index]:
                max_index = j

        a[i], a[max_index] = a[max_index], a[i]

    return a[:k]


scores = [72, 88, 65, 90, 77, 95, 60, 83, 91, 68]

print("Top 5:", top_k_scores(scores, 5))
```

Selection Sort – Minimum Writes

```python
def selection_sort_min_writes(arr):
    a = arr.copy()
    swaps = 0

    for i in range(len(a) - 1):
        min_index = i

        for j in range(i + 1, len(a)):
            if a[j] < a[min_index]:
                min_index = j

        if min_index != i:
            a[i], a[min_index] = a[min_index], a[i]
            swaps += 1

    return a, swaps


result, swaps = selection_sort_min_writes(
    [23.5, 19.2, 25.1, 18.8, 21.4]
)

print("Sorted:", result)
print("Swaps:", swaps)
```

Selection Sort – Library Books

```python
def reorder_shelf(books):
    a = books.copy()
    moves = 0

    for i in range(len(a) - 1):
        min_index = i

        for j in range(i + 1, len(a)):
            if a[j] < a[min_index]:
                min_index = j

        if min_index != i:
            a[i], a[min_index] = a[min_index], a[i]
            moves += 1

    return a, moves


books = [305, 102, 250, 118, 199, 400, 101]

ordered, moves = reorder_shelf(books)

print("Ordered books:", ordered)
print("Physical moves:", moves)
```

Selection Sort – Contest Prize Distribution

```python
def distribute_prizes(participants):
    a = participants.copy()

    for i in range(len(a)):
        max_index = i

        for j in range(i + 1, len(a)):
            if a[j][1] > a[max_index][1]:
                max_index = j

        a[i], a[max_index] = a[max_index], a[i]
        print("Rank", i + 1, ":", a[i][0])

    return a


participants = [
    ("Asha", 88),
    ("Ravi", 95),
    ("Meera", 79),
    ("Dev", 95)
]

ranking = distribute_prizes(participants)
print("Final ranking:", ranking)
```

Selection Sort – Mobile App

```python
import time

def selection_sort(arr):
    a = arr.copy()

    for i in range(len(a) - 1):
        min_index = i

        for j in range(i + 1, len(a)):
            if a[j] < a[min_index]:
                min_index = j

        a[i], a[min_index] = a[min_index], a[i]

    return a


items = [499, 129, 899, 45, 275, 60, 310, 150]

start = time.perf_counter()
result = selection_sort(items)
end = time.perf_counter()

print("Sorted:", result)
print("Time:", end - start)
```
