Insertion Sort – Live Leaderboard

```python
def insert_updated_score(board, score):
    a = board.copy()
    shifts = 0

    i = 0

    while i < len(a) and a[i] > score:
        i += 1

    a.append(score)

    j = len(a) - 1

    while j > i:
        a[j] = a[j - 1]
        j -= 1
        shifts += 1

    a[i] = score

    return a, shifts


board = [980, 875, 760, 690, 500]

result, shifts = insert_updated_score(board, 820)

print("Updated leaderboard:", result)
print("Shifts:", shifts)
```

Insertion Sort – Card Sorting

```python
def pick_up_card(hand, card):
    a = hand.copy()
    i = len(a) - 1

    a.append(card)

    while i >= 0 and a[i] > card:
        a[i + 1] = a[i]
        i -= 1

    a[i + 1] = card

    return a


hand = []

for card in [7, 2, 9, 4, 1]:
    hand = pick_up_card(hand, card)
    print("Hand:", hand)
```

Insertion Sort – Stock Price Feed

```python
def insert_price(prices, price):
    a = prices.copy()
    i = len(a) - 1

    a.append(price)

    while i >= 0 and a[i] > price:
        a[i + 1] = a[i]
        i -= 1

    a[i + 1] = price

    return a


prices = []

for p in [102.5, 98.3, 105.1, 100.0, 97.8]:
    prices = insert_price(prices, p)
    print(prices)

print("Minimum:", prices[0])
print("Maximum:", prices[-1])
```

Insertion Sort – Playlist Reordering

```python
def insert_song(playlist, song):
    a = playlist.copy()
    i = len(a) - 1

    a.append(song)

    while i >= 0 and a[i][1] > song[1]:
        a[i + 1] = a[i]
        i -= 1

    a[i + 1] = song

    return a


playlist = [
    ("Intro", 120),
    ("Chill Beat", 210),
    ("Long Jam", 340)
]

playlist = insert_song(playlist, ("Quick Track", 180))

print("Playlist:")
for song in playlist:
    print(song)
```

Insertion Sort – Sensor Log

```python
def insertion_sort_count_shifts(arr):
    a = arr.copy()
    shifts = 0

    for i in range(1, len(a)):
        key = a[i]
        j = i - 1

        while j >= 0 and a[j] > key:
            a[j + 1] = a[j]
            j -= 1
            shifts += 1

        a[j + 1] = key

    return a, shifts


log = [18.2, 18.5, 18.9, 17.9, 19.1, 19.4, 19.0]

result, shifts = insertion_sort_count_shifts(log)

print("Sorted log:", result)
print("Shifts:", shifts)
```
