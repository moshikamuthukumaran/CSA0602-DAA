Coin Change – Supermarket Billing

```python
def coin_change(coins, amount):
    coins.sort(reverse=True)
    result = []

    for coin in coins:
        while amount >= coin:
            amount -= coin
            result.append(coin)

    return result

coins = [1, 2, 5, 10]
amount = 28

result = coin_change(coins, amount)
print(*result)
print("Minimum coins =", len(result))
```

Fractional Knapsack – Food Delivery Truck

```python
def fractional_knapsack(weights, profits, capacity):
    items = sorted(zip(weights, profits),
                   key=lambda x: x[1] / x[0],
                   reverse=True)

    total = 0

    for w, p in items:
        if capacity >= w:
            capacity -= w
            total += p
        else:
            total += p * capacity / w
            break

    return total

weights = [10, 20, 30]
profits = [60, 100, 120]
capacity = 50

print("Maximum profit =", fractional_knapsack(weights, profits, capacity))
```

Job Sequencing – Software Company Tasks

```python
def job_sequencing(jobs, deadlines, profits):
    data = sorted(zip(jobs, deadlines, profits),
                  key=lambda x: x[2], reverse=True)

    slots = [None] * (max(deadlines) + 1)
    total = 0

    for job, deadline, profit in data:
        for i in range(deadline, 0, -1):
            if slots[i] is None:
                slots[i] = job
                total += profit
                break

    return [x for x in slots if x], total

jobs = ['A', 'B', 'C', 'D']
deadlines = [2, 1, 2, 1]
profits = [100, 19, 27, 25]

selected, profit = job_sequencing(jobs, deadlines, profits)

print("Selected jobs =", *selected)
print("Maximum profit =", profit)
```

Huffman Coding – Text Compression

```python
import heapq

def huffman(chars, freq):
    heap = [[f, c] for c, f in zip(chars, freq)]
    heapq.heapify(heap)

    codes = {c: "" for c in chars}

    while len(heap) > 1:
        a = heapq.heappop(heap)
        b = heapq.heappop(heap)

        for c in a[1:]:
            if isinstance(c, str):
                codes[c] = "0" + codes[c]

        for c in b[1:]:
            if isinstance(c, str):
                codes[c] = "1" + codes[c]

        heapq.heappush(heap, [a[0] + b[0]] + a[1:] + b[1:])

    return codes

chars = ['a', 'b', 'c', 'd', 'e', 'f']
freq = [5, 9, 12, 13, 16, 45]

codes = huffman(chars, freq)

for c, code in sorted(codes.items(), key=lambda x: x[1]):
    print(c + ":", code)
```
